# PDF to Educational Video — Android launcher

This notebook gives you a phone-friendly upload screen. Run the two cells in order.

## Before you run it

1. Open the **Secrets** panel using the key icon on the left.
2. Add a secret named exactly `OPENAI_API_KEY`.
3. Paste your API key as the value and turn **Notebook access** on.
4. Never paste your key into a code cell, GitHub, a screenshot, or chat.


```python
# STEP 1 — Install the project (usually 2-4 minutes)
import pathlib
import shutil
import subprocess

repo_url = 'https://github.com/jromjenine76-alt/pdf-to-video-converter.git'
project_dir = pathlib.Path('/content/pdf-to-video-converter')

if project_dir.exists():
    subprocess.run(['git', '-C', str(project_dir), 'pull', '--ff-only'], check=True)
else:
    subprocess.run(['git', 'clone', '--depth', '1', repo_url, str(project_dir)], check=True)

if shutil.which('ffmpeg') is None:
    subprocess.run(['apt-get', '-qq', 'update'], check=True)
    subprocess.run(['apt-get', '-qq', 'install', '-y', 'ffmpeg'], check=True)

subprocess.run(['python', '-m', 'pip', 'install', '-q', '-r', str(project_dir / 'requirements.txt')], check=True)
print('✅ Setup complete. Run STEP 2 below.')
```

## Start the private converter

The next cell prints a random PIN and creates a temporary web link. The link is protected by that PIN, but do not share either one. Keep this cell running while you make the video.


```python
# STEP 2 — Launch the phone-friendly converter
import os
import secrets
import sys
from google.colab import userdata

try:
    saved_key = (userdata.get('OPENAI_API_KEY') or '').strip()
except Exception:
    saved_key = ''

if not saved_key:
    raise ValueError('OPENAI_API_KEY is missing. Add it in the Colab Secrets panel and enable notebook access.')
del saved_key

project_dir = '/content/pdf-to-video-converter'
os.chdir(project_dir)
if project_dir not in sys.path:
    sys.path.insert(0, project_dir)

from app import APP_CSS, APP_THEME, build_demo

pin = f'{secrets.randbelow(900000) + 100000}'
print('🔐 Username: video')
print('🔐 Your private PIN:', pin)
print('Open the public Gradio link that appears below, then use that username and PIN.')

demo = build_demo().queue(default_concurrency_limit=1)
demo.launch(share=True, auth=('video', pin), show_error=False, debug=True, theme=APP_THEME, css=APP_CSS)
```
